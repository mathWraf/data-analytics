# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(part_code):
    return part_code[:part_code.find(":")]

def get_product_num(part_code):
    return part_code[part_code.find(":")+1:part_code.find("-")]

def get_size(part_code):
    return part_code[part_code.find("-")+1:]
```

```python
code_1 = "ACME:123-L"

print(f"For {code_1}, the supplier is {get_supplier_code(code_1)}. Part number {get_product_num(code_1)} is available in size {get_size(code_1)}.")
```

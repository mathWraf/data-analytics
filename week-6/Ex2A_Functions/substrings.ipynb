## Lab 1. Working with names – parsing strings

```python
test_name = "Joe Scott"
```

```python
def parse_and_display(name):
    for i,letter in enumerate(name):
        if letter == " ":
            first = name[:i]
            last = name[i+1:]
            print(f"Full Name: {name}")
            print(f"First Name: {first}")
            print(f"Last Name: {last}")
            
```

```python
parse_and_display(test_name)
```

```python
test_list = ["Grace Flores", "JB Oinonen", "Ken Penn"]
```

```python
parse_and_display(test_list[0])
```

```python
parse_and_display(test_list[1])
```

```python
parse_and_display(test_list[2])
```

```python
parse_and_display("Matthew Pena")
```

```python
def parse_and_display2(name):
    name = name.split()
    if len(name) > 2:
        first = name[0]
        middle = name[1]
        last = name[2]
        
        print(f"Full Name: {name[0]} {name[1]} {name[2]}")
        print(f"First Name: {first}")
        print(f"Middle Name: {middle}")
        print(f"Last Name: {last}")
    elif len(name) > 1:
        first = name[0]
        last = name[1]
        print(f"Full Name: {name[0]} {name[1]}")
        print(f"First Name: {first}")
        print(f"Last Name: {last}")
    else:
        first = name[0]
        print(f"Full Name: {name[0]}")
        print(f"First Name: {first}")
    
```

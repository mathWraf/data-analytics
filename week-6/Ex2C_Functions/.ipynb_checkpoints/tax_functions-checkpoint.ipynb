Creating custom functions to calculate taxes

```python
def get_soc_sec_tax(gross_pay):
    return 0.062 * gross_pay
```

```python
def get_medicare_tax(gross_pay):
    return 0.0145 * gross_pay
```

```python
def get_federal_tax(gross_pay, withholding_code):
    
    if withholding_code > 4:
        return (0.18-((withholding_code - 4) * 0.005)) * gross_pay
    elif withholding_code == 4:
        return 0.18 * gross_pay
    elif withholding_code == 3:
        return 0.185 * gross_pay
    elif withholding_code == 2:
        return 0.195 * gross_pay
    elif withholding_code == 1:
        return 0.21 * gross_pay
    else:
        return 0.23 * gross_pay
```

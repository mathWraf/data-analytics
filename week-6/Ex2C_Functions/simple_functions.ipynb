defining custom functions

```python
def favorite_things(name,movie,music):
    return f"Name: {name}\nFavorite Movie: {movie}\nFevorite musician/band: {music}"
```

```python
print(favorite_things("John","Dune","Bruno Mars"))
```

```python
def why_im_here(purpose):
    return f"Why I joined the Year Up United Data Analyst Program: {purpose}"
```

```python
print(why_im_here("to develop tech skills and gain valuable experience in the tech industry"))
```

```python
def favorite_place(place,why):
    return f"One of my favorite places to visit is {place} because {why}"
```

```python
print(favorite_place("Shopping Mall", "there's a lot of stuff to do there."))
```

```python
print(favorite_place("Japanese Restaurants", "I like to eat ramen."))
```

Creating custom functions to calculate taxes

```python
def get_soc_sec_tax(gross_pay):
    return 0.062 * gross_pay
```

```python
def get_medicare_tax(gross_pay):
    return 0.0145 * gross_pay
```

```python
def get_federal_tax(gross_pay, withholding_code):
    
    if withholding_code > 4:
        return (0.18-((withholding_code - 4) * 0.005)) * gross_pay
    elif withholding_code == 4:
        return 0.18 * gross_pay
    elif withholding_code == 3:
        return 0.185 * gross_pay
    elif withholding_code == 2:
        return 0.195 * gross_pay
    elif withholding_code == 1:
        return 0.21 * gross_pay
    else:
        return 0.23 * gross_pay
```

```python
#Person 1
print(f"With a gross pay of $750, the Social Security tax is ${get_soc_sec_tax(750):,.2f}")
print(f"With a gross pay of $750, the Medicare tax is ${get_medicare_tax(750):,.2f}")
print(f"With a gross pay of $750, the Federal tax is ${get_federal_tax(750, 0):,.2f}")
```

```python
#Person 2
print(f"With a gross pay of $1550, the Social Security tax is ${get_soc_sec_tax(1550):,.2f}")
print(f"With a gross pay of $1550, the Medicare tax is ${get_medicare_tax(1550):,.2f}")
print(f"With a gross pay of $1550, the Federal tax is ${get_federal_tax(1550, 2):,.2f}")
```
